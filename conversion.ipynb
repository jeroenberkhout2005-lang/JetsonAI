```python
import torch
import torch.nn as nn

# Example: replace with your actual model class
class ParticleNet(nn.Module):
    def __init__(self):
        super(ParticleNet, self).__init__()
        self.conv1 = nn.Conv2d(1, 16, 3, padding=1)
        self.relu = nn.ReLU()
        self.conv2 = nn.Conv2d(16, 1, 3, padding=1)

    def forward(self, x):
        x = self.relu(self.conv1(x))
        x = self.conv2(x)
        return x

# Load your model
model = ParticleNet()
checkpoint = torch.load("yt_model.pth", map_location="cpu")
model.load_state_dict(checkpoint, strict=False)
model.eval()

# Create a dummy input with the correct shape
# For example: 1 channel, 256x256 image, batch size 1
dummy_input = torch.randn(1, 1, 256, 256)

# Export to ONNX
torch.onnx.export(
    model,
    dummy_input,
    "yt_model.onnx",
    input_names=["input"],
    output_names=["output"],
    opset_version=11,
    dynamic_axes={"input": {0: "batch_size"}, "output": {0: "batch_size"}}
)

print("ONNX export complete: yt_model.onnx")
```
